**CREATING OF MATRIX GRASCALE WITH DEEP LEARNING**

```python
import matplotlib as plt
import numpy as np
import pandas as pd
import PIL #images
import pytest
import sklearn
import scipy
import seaborn as sns
import tensorflow as sns
import matplotlib.pyplot as plt
from matplotlib.pyplot import rcParams
```

```python
bw= np.random.binomial(1,0.5,size=(10,10))
bw #creating a array with random.binomial function to randomly generate binary values
```

```python
plt.imshow(bw, cmap='gray', vmin=0, vmax=1)  #using imshow to show image
plt.title("10×10 Black & White Image")
plt.show()
```

```python
#grayscale images have the range 0-256, lets generate an image with random matrix of this
gs=np.random.randint(0,256, size=(10,10))
gs
```

```python
plt.imshow(gs, cmap='gray')  #using imshow to show image
plt.title("10×10 Random Grayscale image")
plt.show()
```
