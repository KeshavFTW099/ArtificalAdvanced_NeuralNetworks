```python
import tensorflow as tf
import matplotlib.pyplot as plt

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
```

```python
print("Training Images :", x_train.shape)
print("Training Labels :", y_train.shape)

print("Testing Images  :", x_test.shape)
print("Testing Labels  :", y_test.shape)
```

```python
# Checking the shape of the training and testing datasets
print("Training Images :", x_train.shape)
print("Testing Images :", x_test.shape)
```

```python
first_img = x_train[0]

print(first_img)

plt.imshow(first_img, cmap='gray')
plt.title(f"Label: {y_train[0]}")
plt.show()
```

```python
# Flattening the 28x28 images into 784-dimensional vectors
x_train_flat = x_train.reshape(60000, 784)
print("x_train_flat shape :", x_train_flat.shape)
```

```python
x_test_flat = x_test.reshape(10000, 784)
print("x_test_flat shape :", x_test_flat.shape)
```

```python
# Checking the minimum and maximum pixel values
print("Minimum Pixel Value :", x_train_flat.min())
print("Maximum Pixel Value :", x_train_flat.max())
```

```python
# Neural networks are sensitive to the absolute size of inputs,
# so convert the data type to float32
x_train_sc = x_train_flat.astype('float32')
x_test_sc = x_test_flat.astype('float32')
```

```python

# Normalize pixel values to the range 0-1
x_train_sc = x_train_sc / 255.0
x_test_sc = x_test_sc / 255.0
```

```python
# Checking the normalized pixel values
print("Minimum after normalization :", x_train_sc.min())
print("Maximum after normalization :", x_train_sc.max())
```

```python

# Convert classes into One-Hot Encoding
from tensorflow.keras.utils import to_categorical

y_train_cat = to_categorical(y_train)
y_test_cat = to_categorical(y_test)

print("y_train_cat shape :", y_train_cat.shape)
print("y_test_cat shape :", y_test_cat.shape)

print("Original Label :", y_train[0])
print("One-Hot Encoded Label :", y_train_cat[0])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Creating the Sequential Model
model = Sequential()

# Input Layer + Hidden Layer 1
model.add(Dense(512, activation='relu', input_shape=(784,)))

# Hidden Layer 2
model.add(Dense(256, activation='relu'))

# Hidden Layer 3
model.add(Dense(128, activation='relu'))

# Output Layer
model.add(Dense(10, activation='softmax'))
```

```python
model.compile(
    loss='categorical_crossentropy',
    optimizer='rmsprop',
    metrics=['accuracy']
)
```
